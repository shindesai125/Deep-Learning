```python
!pip install wikipedia-api
import wikipediaapi

wiki = wikipediaapi.Wikipedia(user_agent='MyWikipediaApp/1.0 (myemail@example.com)', language='en')
page=wiki.page("Artificial Intelligence")
text = page.text
print(text[:1000])
```

```python
text = text[:len(text)//8]
```

```python
text
```

```python
import tensorflow as tf
from tensorflow.keras.preprocessing.text import Tokenizer
tokenizer = Tokenizer()
```

```python
tokenizer.fit_on_texts([text])
```

```python
tokenizer.word_index
```

```python
len(tokenizer.word_index)
```

```python
input_sequences=[]
for sentences in text.split('\n'):
  tokenized_sentence = tokenizer.texts_to_sequences([text])[0]

  for i in range(1,len(tokenized_sentence)):
    input_sequences.append(tokenized_sentence[: i+1])
```

```python
input_sequences
```

```python
max_len= max([len(x) for x in input_sequences])
max_len
```

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences
padded_input_sequences = pad_sequences(input_sequences, maxlen = max_len, padding='pre')
```

```python
padded_input_sequences
```

```python
X = padded_input_sequences[:,:-1]
y = padded_input_sequences[:,-1]
```

```python
X
```

```python
y
```

```python
print("The shape of Input(X) is", X.shape)
print("The shape of Output(y) is", y.shape)
```

```python
from tensorflow.keras.utils import to_categorical
y = to_categorical(y, num_classes = 682)
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense
```

```python
model = Sequential()
model.add(Embedding(682, 100, input_length = 1674))
model.add(LSTM(100))
model.add(Dense(682, activation = 'softmax'))

# Force build
model.build(input_shape=(None, max_len))
```

```python
model.summary()
```

```python
model.compile(loss = 'categorical_crossentropy', optimizer = 'adam', metrics = ['accuracy'])
```

```python
model.fit(X, y, epochs=10)
```

```python
import numpy as np
text = "High-profile applications"

for i in range(10):
  # tokenize
  tokenized_text = tokenizer.texts_to_sequences([text])[0]

  # padding
  padded_tokenized_text = pad_sequences([tokenized_text], maxlen = 1673, padding='pre')

  position=np.argmax(model.predict(padded_tokenized_text))



  for word,index in tokenizer.word_index.items():
    if index ==position:
      text=text+" "+ word
      print(text)
```
